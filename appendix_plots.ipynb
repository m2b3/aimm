```python
%load_ext autoreload
%autoreload 2

from pathlib import Path

import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec
from PIL import Image
```

## Prompt example images

The same five images `2_run_model.ipynb` passes to the model as
`reference_imgs`, laid out as they are described in the prompt.

```python
PROMPT_IMG_DIR = Path('docs/original_prompt')

example_scene = Image.open(PROMPT_IMG_DIR / 'example_scene.png').convert('RGB')
low1  = Image.open(PROMPT_IMG_DIR / 'low_meaning_example1.png').convert('RGB')
low2  = Image.open(PROMPT_IMG_DIR / 'low_meaning_example2.png').convert('RGB')
high1 = Image.open(PROMPT_IMG_DIR / 'high_meaning_example1.png').convert('RGB')
high2 = Image.open(PROMPT_IMG_DIR / 'high_meaning_example2.png').convert('RGB')

plt.rcParams['font.family'] = 'Arial'

fig = plt.figure(figsize=(8, 3.5), facecolor='white')
gs_app = gridspec.GridSpec(
    2, 3, figure=fig,
    width_ratios=[2, 1, 1],
    wspace=0.06, hspace=0.2,
    left=0.02, right=0.98, bottom=0.06, top=0.96
)

ax_scene = fig.add_subplot(gs_app[:, 0])
ax_low1  = fig.add_subplot(gs_app[0, 1])
ax_low2  = fig.add_subplot(gs_app[0, 2])
ax_high1 = fig.add_subplot(gs_app[1, 1])
ax_high2 = fig.add_subplot(gs_app[1, 2])

panels = [
    (ax_scene, example_scene, 'Example scene'),
    (ax_low1,  low1,          'Low-meaning example 1'),
    (ax_low2,  low2,          'Low-meaning example 2'),
    (ax_high1, high1,         'High-meaning example 1'),
    (ax_high2, high2,         'High-meaning example 2'),
]
for ax, im, title in panels:
    ax.imshow(im)
    ax.axis('off')
    ax.set_title(title, fontsize=11, pad=5)

plt.savefig('results/paper_plots/appendix_prompt_examples.png', dpi=300,
            bbox_inches='tight', facecolor='white')
plt.show()
print('Saved to results/paper_plots/appendix_prompt_examples.png')
```
